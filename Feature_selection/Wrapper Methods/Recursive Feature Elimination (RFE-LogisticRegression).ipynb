```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFE
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
```

```python
#Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data/Labelled_data_set.csv')
Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data2/Data.csv')
```

```python
Data = Data.iloc[:, 1:]
#Data = Data.iloc[:, 0:]
#Data = Data.iloc[:, 1:-14]
#Data = Data.iloc[:, 8:-1]
```

```python
Data
```

```python
Data['Labels'] = Data['FPS'].apply(lambda x: 0 if x>=60 else 1)
#Data["Labels"] = Data["MotionToPhotonLatency"].apply(lambda x: 0 if x <= 20 else 1)
```

```python
Data.shape
```

```python
X = Data.drop('Labels', axis = 1)
```

```python
y = Data['Labels']
```

```python
feature_names = X.columns
```

```python
# Splitting the dataset into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Defining the model
model = LogisticRegression(solver='liblinear', max_iter=1000)
```

```python
# Defining RFE
n_features_to_select = 5
rfe = RFE(estimator=model, n_features_to_select=n_features_to_select)
```

```python
# Fitting RFE
rfe.fit(X_train, y_train)
```

```python
# Getting the selected features
selected_features = np.where(rfe.support_)[0]
selected_feature_names = feature_names[selected_features]
```

```python
# Creating a DataFrame with selected features and their names
selected_features_df = pd.DataFrame({'Feature index': selected_features, 'Feature Name': selected_feature_names })
```

```python
selected_features_df
```

```python
selected_feature_names
```

```python
# Evaluating the performance of the model with the selected features
X_train_rfe = rfe.transform(X_train)
X_test_rfe = rfe.transform(X_test)
```

```python
model.fit(X_train_rfe, y_train)
```

```python
y_pred = model.predict(X_test_rfe)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# Transforming the dataset with the selected features
X_rfe = rfe.transform(X)
```

```python
# Evaluating the performance of the model using k-fold cross-validation
k = 10  # Number of folds
scores = cross_val_score(model, X_rfe, y, cv=k, scoring='accuracy')
```

```python
# Printing the cross-validation scores and average accuracy
print(f"Cross-validation scores ({k}-fold):", scores)
print("Average accuracy:", np.mean(scores))
```
