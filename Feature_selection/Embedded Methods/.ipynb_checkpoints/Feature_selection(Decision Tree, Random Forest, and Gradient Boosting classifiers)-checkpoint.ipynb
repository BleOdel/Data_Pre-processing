```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load your dataset
#Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data/Labelled_data_set.csv')
Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data2/Data.csv')
```

```python
Data
```

```python
Data.info()
```

```python
Data = Data.iloc[:, 1:]
#Data = Data.iloc[:, 0:]
#Data = Data.iloc[:, 1:-14]
#Data = Data.iloc[:, 8:-1]
```

```python
Data
```

```python
Data['Labels'] = Data['FPS'].apply(lambda x: 0 if x>=60 else 1)
#Data["Labels"] = Data["MotionToPhotonLatency"].apply(lambda x: 0 if x <= 20 else 1)
```

```python
Data
```

```python
Data.shape
```

```python
X = Data.drop('Labels', axis = 1)
```

```python
y = Data['Labels']
```

```python
feature_names = X.columns
```

```python
# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

```python
# Train and evaluate the classifiers
dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)
dt_pred = dt_classifier.predict(X_test)
dt_accuracy = accuracy_score(y_test, dt_pred)

rf_classifier = RandomForestClassifier(random_state=42)
rf_classifier.fit(X_train, y_train)
rf_pred = rf_classifier.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)


gb_classifier = GradientBoostingClassifier(random_state=42)
gb_classifier.fit(X_train, y_train)
gb_pred = gb_classifier.predict(X_test)
gb_accuracy = accuracy_score(y_test, gb_pred)
```

```python
# Plot the accuracy comparison
accuracy_df = pd.DataFrame({
    'Classifier': ['Decision Tree', 'Random Forest', 'Gradient Boosting'],
    'Accuracy': [dt_accuracy, rf_accuracy, gb_accuracy]
})
```

```python
accuracy_df
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Classifier', y='Accuracy', data=accuracy_df)
plt.title("Classifier Accuracy Comparison")
plt.xlabel("Classifier")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.show()
```

```python
dt_importances = dt_classifier.feature_importances_
dt_feature_importances_df = pd.DataFrame({"Feature": feature_names, "Importance": dt_importances})
dt_feature_importances_df.sort_values(by="Importance", ascending=False, inplace=True)

rf_importances = rf_classifier.feature_importances_
rf_feature_importances_df = pd.DataFrame({"Feature": feature_names, "Importance": rf_importances})
rf_feature_importances_df.sort_values(by="Importance", ascending=False, inplace=True)

gb_importances = gb_classifier.feature_importances_
gb_feature_importances_df = pd.DataFrame({"Feature": feature_names, "Importance": gb_importances})
gb_feature_importances_df.sort_values(by="Importance", ascending=False, inplace=True)
```

```python
def plot_feature_importances(feature_importances_df, classifier_name):
    plt.figure(figsize=(12, 6))
    sns.barplot(x="Feature", y="Importance", data=feature_importances_df)
    plt.xticks(rotation=90)
    plt.title(f"{classifier_name} Feature Importances")
    plt.xlabel("Features")
    plt.ylabel("Importance")
    plt.show()
```

```python
plot_feature_importances(dt_feature_importances_df, "Decision Tree")
plot_feature_importances(rf_feature_importances_df, "Random Forest")
plot_feature_importances(gb_feature_importances_df, "Gradient Boosting")
```
