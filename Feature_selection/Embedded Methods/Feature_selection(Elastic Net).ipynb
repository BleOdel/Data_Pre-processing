```python
import numpy as np
import pandas as pd
from sklearn.linear_model import ElasticNetCV, ElasticNet
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load your dataset
#Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data/Labelled_data_set.csv')
Data = pd.read_csv('/Users/blessingodeleye/Documents/VR_data2/Data.csv')
```

```python
Data = Data.iloc[:, 1:]
#Data = Data.iloc[:, 0:]
#Data = Data.iloc[:, 1:-14]
#Data = Data.iloc[:, 8:-1]
```

```python
Data['Labels'] = Data['FPS'].apply(lambda x: 0 if x>=60 else 1)
#Data["Labels"] = Data["MotionToPhotonLatency"].apply(lambda x: 0 if x <= 20 else 1)
```

```python
Data
```

```python
Data.shape
```

```python
X = Data.drop('Labels', axis = 1)
```

```python
y = Data['Labels']
```

```python
feature_names = X.columns
```

```python
# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled , y, test_size=0.3, random_state=42)
```

```python
# Using ElasticNetCV for cross-validated Elastic Net
model_cv = ElasticNetCV(alphas=np.logspace(-5, 1, 100), l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9], cv=5)
model_cv.fit(X_train, y_train)
best_alpha_cv = model_cv.alpha_
best_l1_ratio_cv = model_cv.l1_ratio_
```

```python
print(f"Best alpha using ElasticNetCV: {best_alpha_cv:.6f}")
print(f"Best l1_ratio using ElasticNetCV: {best_l1_ratio_cv:.2f}")
```

```python
# Create the Elastic Net model with the best alpha and l1_ratio values
model = ElasticNet(alpha=best_alpha_cv, l1_ratio=best_l1_ratio_cv, random_state=42)
```

```python
# Fit the model to the training data
model.fit(X_train, y_train)
```

```python
# Predict on the test set
y_pred = model.predict(X_test)
```

```python
# Evaluate the model using mean squared error (MSE)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print(f"Test set RMSE: {rmse:.4f}")
```

```python
# Get the coefficients of the Elastic Net model
coefficients = model.coef_
```

```python
# Display the coefficients and corresponding feature names
elastic_net_coefficients = pd.DataFrame(coefficients, index=feature_names, columns=["Coefficients"])
```

```python
elastic_net_coefficients
```

```python
# Reset index to convert feature names to a column
elastic_net_coefficients.reset_index(inplace=True)
elastic_net_coefficients.rename(columns={'index': 'Feature'}, inplace=True)
```

```python
# Sort the DataFrame by coefficient values (from highest to lowest)
elastic_net_coefficients.sort_values(by='Coefficients', ascending=False, inplace=True)
```

```python
# Print the sorted DataFrame
print("Sorted Elastic Net coefficients:")
print(elastic_net_coefficients)
```

```python
# Plot the bar plot
plt.figure(figsize=(12, 6))
sns.barplot(x='Feature', y='Coefficients', data=elastic_net_coefficients)
plt.xticks(rotation=90)
plt.title("Elastic Net Regression Coefficients (Sorted)")
plt.xlabel("Features")
plt.ylabel("Coefficient")
plt.show()
```

```python
# Identify selected features based on non-zero coefficients
selected_features = feature_names[np.nonzero(coefficients)]
print("\nSelected features:")
print(selected_features)
```

```python
#Extract the non-zero coefficients
non_zero_coefficients = coefficients[np.nonzero(coefficients)]
```

```python
# Create a DataFrame with the non-zero coefficients and corresponding feature names
elastic_net_nonzerocoef = pd.DataFrame({"Feature": selected_features, "Coefficient": non_zero_coefficients})
```
