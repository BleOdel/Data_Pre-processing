```python
import pandas as pd
import numpy as np
```

```python
# Load the CSV data
csv_data = "/Users/blessingodeleye/Documents/VR_data2/FM_FS_Data_set.csv" # Replace this with the path to your CSV file
# Load data from the CSV file
data = pd.read_csv(csv_data, index_col=0)
```

```python
data
```

```python
# Transpose the data to make it easier to work with
data = data.T
```

```python
data
```

```python
# Choose the number of top features you want to select (e.g., 5)
N = 6
```

```python
# Extract the top N features from each model
top_n_features = []
for _, row in data.iterrows():
    sorted_row = row.sort_values(ascending=True)
    top_n_features.extend(sorted_row.head(N).index.tolist())
```

```python
# Count the occurrences of each feature
feature_counts = pd.Series(top_n_features).value_counts()
```

```python
# Choose the features with the highest occurrence counts
selected_features = feature_counts.head(N).index.tolist()
```

```python
print("Selected features:", selected_features)
```

```python
df = pd.DataFrame({'Selected features' : selected_features})
```
